# DocLens experiment analysis

Load saved evaluation reports from `experiments/results/` and compare field accuracy across strategies.

```python
from pathlib import Path
import json
import pandas as pd

results = Path("../experiments/results")
rows = []
for path in sorted(results.glob("*.json")):
    report = json.loads(path.read_text())
    row = {
        "file": path.name,
        "strategy": report["strategy"],
        "model": report["model"],
        "accuracy": report["accuracy"],
        "latency_ms": report["latency_ms_mean"],
        "incorrect": report["incorrect"],
        "missing": report["missing"],
    }
    row.update({f"acc_{k}": v for k, v in report.get("field_accuracy", {}).items()})
    rows.append(row)
pd.DataFrame(rows)
```
